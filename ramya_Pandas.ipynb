Q1. A student assessment dataset contains missing marks, duplicate student records, inconsistent department names such as IT, it, and I.T., and marks stored as text. Develop a cleaning sequence to standardize the data, handle missing values, remove duplicates, and convert marks into a suitable numeric type. State one risk of deleting missing values without investigating why they are missing. 
Dataset: student_assessment_dirty.csv



1.Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().
import pandas as pd
df = pd.read_csv("students.csv")
print(df.head())
print(df.info())
print(df.isnull().sum())

we load the CSV file using read_csv(). Then we use head(), info(), and isnull().sum() to understand the data, its columns, data types, and missing values.

2.Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.
df["Department"] = df["Department"].str.strip().str.upper()
df["Department"] = df["Department"].replace({
    "I.T.": "IT"
})
Some department names are written differently, like IT, it, and I.T.. We convert them into one common format so Pandas treats them as the same department.

3.identify and remove exact duplicate student assessment records.

before = len(df)
df = df.drop_duplicates()
removed = before - len(df)
print("Records removed:", removed)

The dataset contains the same student assessment record more than once. We use drop_duplicates() to remove the exact duplicate and avoid counting the same record twice.

4.Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.

df["Mark_Status"] = df["Marks"].astype(str).str.strip()
df["Marks"] = pd.to_numeric(
    df["Marks"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce"
)
The Marks column contains values like 88 marks, Absent, NA, and blank values. We convert valid marks into numbers and treat values that are not actual marks as missing values.

5.Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.

print(df[df["Marks"].isnull()][["Student_ID", "Student_Name", "Mark_Status"]])

We first check why the marks are missing instead of directly deleting them. Absent, NA, and blank values can have different meanings, so deleting them without checking can result in loss of useful information.

6.Identify marks outside the valid range of 0–100 and decide how they should be handled

invalid = df[(df["Marks"] < 0) | (df["Marks"] > 100)]
print(invalid[["Student_ID", "Student_Name", "Marks"]])

Marks should normally be between 0 and 100. The value 105 is outside this range, so we identify it as an invalid value and investigate it before deciding whether to correct or remove it.


7.Display the cleaned DataFrame and report the number of records removed, missing marks remaining, and invalid marks identified.

print(df)
print("Records removed:", removed)
print("Missing marks remaining:", df["Marks"].isnull().sum())
print("Invalid marks identified:", len(invalid))

we display the cleaned DataFrame and report how many duplicate records were removed, how many missing marks remain, and how many invalid marks were found.
