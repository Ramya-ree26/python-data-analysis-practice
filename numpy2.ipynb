```python
import numpy as np
rng=np.random .default_rng(42)
print("random number")
print(rng.integers(1,10,size=5))
print("random floats")
print(rng.random(5))
print("normal distribution")
print(rng.normal(loc=0,scale=1,size=5))
x=rng.random(((1,2,4)))
print(x)
print("Shape",x.shape)
```

```python
students=np.array(["a","b","c","d"])
print("one student",rng.choice(students))
print("three students",rng.choice(students,size=3,replace=false))
print("two students",rng.choice(students,size=2,replace=false))
```

```python
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
print("Intersection :",np.intersect1d(a,b))
print("Union:",np.union1d(a,b))
print("A-B:",np.setdiff1d(a,b))
print("Superset:",np.all(np.isin(a,b)))




```

```python
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
print("element wise multiplication")
print(a*b)
print("matrix multiplication")
print(a@b)
print(" Dot product")
print(np.dot(a,b))
```

```python
A=np.array([[1,2],[4,5]],dtype=float)
print("Determinant",np.linalg.det(A))
print("inverse",np.linalg.inv(A))
print("\n A @ inverse:")
print(A @ np.linalg.inv(A))

```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
print("Concatenate",np.concatenate([a,b]))
print("Stack",np.stack([a,b]))
```

```python
x=np.array([[1,2],[3,4]])
y=np.array([[5,6],[7,8]])
print("vertical stack")
print(np.vstack([x,y]))
print("horizontal stack")
print(np.hstack([x,y]))
```
