```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print("Matplotlib version:",plt.matplotlib.__version__)
print("Pandas version",pd.__version__)
```

```python
x=np.linspace(0,10,100)
y=np.sin(x)

fig, ax = plt.subplots(figsize=(8,6))
ax.plot(x, y)
ax.set_title("Object oriented interface: Figure + plots")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
plt.show()

print("Type of fig:", type(fig))
print("Type of ax:", type(ax))
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,y)
plt.title("pyplot interface (same result)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(x,y)
plt.title("pyplot interface (same result)")
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.show()
```

```python
df=pd.read_csv("C:\\Users\\ABC\\Downloads\\web_traffic_daily.csv")
df.head()
```

```python
plt.figure(figsize=(10,4))
plt.plot(df["date"],df["visits"],color="tab:blue",linewidth=1.5,label="Visits")
plt.title("Daily visits over one year")
plt.xlabel("Date")
plt.ylabel("Visits")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
ax.set_yscale("log")
```

```python
plt.figure(figsize=(10,4))
plt.plot(df["date"],df["visits"],df["signups"],color="tab:orange",linewidth=1.5,linestyle="--",label="Visits")
plt.title("Daily visits over one year")
plt.xlabel("Date")
plt.ylabel("Visits")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
ax.set_yscale("log")

```

```python
fig, ax = plt.subplots(figsize=(10,6))
ax.plot(df["date"],df["visits"],color="tab:blue",linewidth=1.5,label="Visits",linestyle="--")
ax.plot(df["signups"],color="tab:orange",linewidth=1.5,label="Signups",linestyle="--")
ax.set_title("Daily visits over one year")
ax.set_xlabel("Date")
ax.set_ylabel("Visits")
ax.legend()
ax.set_yscale("log")
```

```python
plt.figure(figsize=(6,5))
sc=plt.scatter(df["page_load_ms"],df["error_count"],c=df["visits"],cmap="viridis",alpha=0.7)
plt.title("Errors vs page load time(color=visits)")
plt.xlabel("Page load time(ms)")
plt.ylabel("Error count")
plt.colorbar(sc,label="visits")
plt.show()
```

```python
plt.figure(figsize=(6,5))
sc=plt.scatter(df["page_load_ms"],df["error_count"],c=df["deployment_flag"],cmap="plasma",alpha=0.7)
plt.title("Errors vs page load time(color=visits)")
plt.xlabel("Page load time(ms)")
plt.ylabel("Error count")
plt.colorbar(sc,label="deployment_flag")
plt.show()
```
