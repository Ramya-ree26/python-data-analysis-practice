```python
work=df.copy()
work["Passed"]=work["Exam_Score"]>=50
work["Total_Score"]=work["Assignment_Score"]+work["Exam_Score"]
work["Average_Score"]=work[["Assignment_Score","Exam_Score"]].mean(axis=1)

display(work[["Name","Assignment_Score","Exam_Score","Total_Score","Average_Score","Passed"]])
    

```

```python
work["Performance_Level"]=np.where(
    work["Exam_Score"]>=85,"Excellent",
    np.where(work["Exam_Score"]>=70,"Good","Needs Improvement"))
display(work[["Name","Exam_Score","Performance_Level"]].head(10))
```

```python
filled=df.copy()
filled["Attendance"]=filled["Attendance"].fillna(
    filled["Attendance"].median()
)

filled["Study_Hours"]=filled["Study_Hours"].fillna(
    filled["Study_Hours"].median()
)
filled["Department"]=filled["Department"].fillna("Unknown")
print("Remaining missing values")
print(filled.isna().sum())

display(filled)
```

```python
clean=df.copy()
clean["Name"]=clean["Name"].str.strip().str.title()
clean["Department"]=clean["Department"].str.strip().str.title()

clean["Department"]=clean["Department"].replace({
    "Ai":"AI",
    "Data science":"Data Science",
    "Python":"Python",
    "cloud":"Cloud" 
})
display(clean[["Name","Department"]].head(15))
```

```python
ai_students=clean[
    clean["Department"].str.contains("AI",case=False,na=False)
    ]
display(ai_students[["Name","Department","Exam_Score"]])
```

```python
date_df =clean.copy()
date_df["Enrollment_Date"]=pd.to_datetime(
    date_df["Enrollment_Date"]

)
date_df["Enrollment_Year"]=date_df["Enrollment_Date"].dt.year
date_df["Enrollment_Month"]=date_df["Enrollment_Date"].dt.month
date_df["Enrollment_Day"]=date_df["Enrollment_Date"].dt.day
display(date_df[["Enrollment_Date","Enrollment_Year","Enrollment_Month","Enrollment_Day"]].head(10))

```

```python
dept_avg=(
    clean.groupby("Department")["Exam_Score"].mean().sort_values(ascending=False))
display(dept_avg)

```

```python
summary=(
    clean.groupby("Department")["Exam_Score"].agg(["count","mean","min","max"])
    .round(2)
)
display(summary)
```

```python
#merge
students=clean[["Student_ID","Department","Name"]].copy()
attendance=clean[["Student_ID","Attendance"]].copy()
merged=pd.merge(
    students,
    attendance,
    on="Student_ID",
    how="inner"
)
display(merged.head())
```
